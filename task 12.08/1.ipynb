```python
g1 = float(input("Введіть рівень готовності 1-го підрозділу: "))
g2 = float(input("Введіть рівень готовності 2-го підрозділу: "))
g3 = float(input("Введіть рівень готовності 3-го підрозділу: "))

average = (g1 + g2 + g3) / 3

print("Середній рівень готовності:", average)
```
