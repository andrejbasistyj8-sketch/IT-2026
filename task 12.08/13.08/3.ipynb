```python
def ociniti_rezultatyvnist(vtraty, chyselnist):
    if chyselnist <= 0:
        return "Помилка: чисельність повинна бути більшою за 0."
    vidsotok = (vtraty / chyselnist) * 100
    if vidsotok >= 50:
        return f"Висока результативність ({vidsotok:.1f}% втрат)."
    elif vidsotok >= 20:
        return f"Середня результативність ({vidsotok:.1f}% втрат)."
    else:
        return f"Низька результативність ({vidsotok:.1f}% втрат)."
vtraty = int(input("Введіть втрати ворога: "))
chyselnist = int(input("Введіть чисельність ворога: "))
print(ociniti_rezultatyvnist(vtraty, chyselnist))
```

```python
def seredni_vtraty (vtraty_za_dni):
    if len(vtraty_za_dni) == 0:
        return 0
    return sum(vtraty_za_dni) / len(vtraty_za_dni)
vtraty = [180, 110, 120, 150, 160]
print("Середні втрати:", seredni_vtraty(vtraty))
```

```python
import random
def змоделювати_напрямок():
    напрямку = ["Київ", "Чернігів", "Одеса", "Львів"]
    return random.choice(напрямку)
print("Напрямок руху:", змоделювати_напрямок())
```
