1. Створити список напрямків, по яких ведеться аналіз.
2. Створити словник із втратами підрозділів та порахувати сумарні втрати.
3. Побудувати множину типів виявленої техніки на основі списку спостережень.
4. Вивести підрозділи, втрати яких перевищують 50 осіб.

```python
directions = ["Куп'янський","Лиманський","Сіверський", "Краматорський","Торецький",]    
print("Напрямки, по яких ведеться аналіз:")
for i, direction in enumerate(directions, start=1):
    print(f"{i}. {direction}")
```

```python
losses = {"Танкові війська": 120,"Мотопіхотні війська": 250,"Десантні війська": 60,}
total_losses = sum(losses.values())
print("Втрати підрозділів:")
for unit, loss in losses.items():
    print(f"{unit}: {loss}")
print(f"\nСумарні втрати: {total_losses}")
```

```python
observations = ["БТР","танк", "дрон","артилерія",]
equipment_types = set(observations)
print("Виявлені типи техніки:")
for equipment in equipment_types:
    print(equipment)
```

```python
losses = {"Артилерія": 80,"Мотопіхотні війська": 250, "Танкові війська": 120,}
print("Підрозділи, втрати яких перевищують 50 осіб:")
for unit, loss in losses.items():
    if loss > 50:
        print(f"{unit}: {loss}")
```
